```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
import lightgbm as lgb
import warnings
warnings.filterwarnings('ignore')

train = pd.read_csv(r'data/train.csv')
test = pd.read_csv(r'data/test.csv')

def preprocess(df):
  #bersihin data(preprocessing) dan buat fitur baru(feature engineering)
  df = df.copy()

  #preprocessing
  #buang kolom tidak informatif
  #sampling_depth_cm hanya berisi "0-20"
  df.drop(columns=['sampling_depth_cm'], inplace=True)

  #mengubah YES/NO menjadi 0/1 (Binary Encoding)
  df['has_band_A_flag'] = (df['has_band_A_spectrum'] == 'YES').astype(int)
  df['has_band_B_flag'] = (df['has_band_B_spectrum'] == 'YES').astype(int)
  df.drop(columns=['has_band_A_spectrum', 'has_band_B_spectrum'], inplace = True)

  # Band A tidak ada yang missing
  # Band A menangkap fisik tanah
  band_a_cols = [f'spectral_band_A_PC_{i}' for i in range (1, 16)]
  df['band_A_mean'] = df[band_a_cols].mean(axis=1)
  df['band_A_std'] = df[band_a_cols].std(axis=1)
  df['band_A_max'] = df[band_a_cols].max(axis=1)
  df['band_A_min'] = df[band_a_cols].min(axis=1)
  df['band_A_mean'] = df['band_A_max'] - df['band_A_min']

  # Band B banyak missing
  # Band B menangkap informasi kimia tanah
  band_b_cols = [f'spectral_band_B_PC_{i}' for i in range (1, 16)]
  df['band_B_mean'] = df[band_b_cols].mean(axis=1)
  df['band_B_std'] = df[band_b_cols].std(axis=1)
  df['band_B_max'] = df[band_b_cols].max(axis=1)
  df['band_B_min'] = df[band_b_cols].min(axis=1)

  # fitur rasio kimia tanah
  # rasio lebih informatif dari nilai mentah
  df['Ca_CEC_ratio'] = df['cation_Ca']/(df['cation_exchange_capacity'] + 1e-6)
  df['Mg_CEC_ratio'] = df['cation_Mg']/(df['cation_exchange_capacity'] + 1e-6)
  df['coarse_fine_ratio'] = df['property_particle_coarse']/(df['property_particle_fine'] + 1e-6)

  # Encode kolom kategorikal menjadi angka
  # LightGBM bisa langsung pakai kolom kategori tanpa one-hot encoding
  cat_cols = [
      'sampling_strategy', 'geo_zone_macro', 'geo_zone_micro', 'geo_zone_meso',
      'land_cover_type', 'biome', 'parent_rock_type'
  ]
  for col in cat_cols:
    df[col] = df[col].astype('category').cat.codes
  return df

train_proc = preprocess(train)
test_proc = preprocess(test)

#Trainning Model
#Kolom target dan kolom yang tidak dipakai pada fitur
target = 'property_organic_content'
drop_cols =['sample_id', 'source_id', target]
feature_cols = [c for c in train_proc.columns if c not in drop_cols]
cat_features = [
    'sampling_strategy', 'geo_zone_macro', 'geo_zone_micro', 'geo_zone_meso',
    'land_cover_type', 'biome', 'parent_rock_type'
]
X = train_proc[feature_cols] #fitur yang di trainning
Y = train_proc[target] #Target yang kita prediksi
X_test = test_proc[feature_cols] #Fitur untuk test

print(f"\nJumlah fitur: {len(feature_cols)}")
print("Fitur:", feature_cols[:5], "...")

# Parameter model LightGBM
# Setiap parameter punya efek tersendiri pada model
params = {
    'objective'       : 'regression',  # Ini adalah masalah regresi (prediksi angka)
    'metric'          : 'rmse',         # Ukur performa dengan RMSE
    'learning_rate'   : 0.03,           # Seberapa besar langkah belajar (kecil = lebih stabil)
    'num_leaves'      : 127,            # Kompleksitas pohon (lebih banyak = lebih detail)
    'min_child_samples': 20,            # Minimal sampel per daun (mencegah overfitting)
    'feature_fraction': 0.8,            # 80% fitur dipakai per pohon (acak)
    'bagging_fraction': 0.8,            # 80% data dipakai per iterasi (acak)
    'bagging_freq'    : 5,              # Frekuensi bagging
    'reg_alpha'       : 0.1,            # Regularisasi L1
    'reg_lambda'      : 0.5,            # Regularisasi L2
    'verbose'         : -1,             # Matikan log yang berisik
    'n_jobs'          : -1,             # Pakai semua CPU
}

#5 fold cross validation
n_splits = 5
kf = KFold(n_splits = n_splits, shuffle=True, random_state = 42)
oof_preds  = np.zeros(len(X)) #Out of Fold prediction (evaluasi)
test_preds = np.zeros(len(X_test)) #prediksi test (rata rata semua fold)
scores = []

print("\n" + "="*50)
print("Mulai training 5-Fold Cross-Validation...")
print("="*50)

for fold, (train_idx, val_idx) in enumerate (kf.split(X)):
  # Pisahkan data training dan validasi
  X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
  Y_tr, Y_val = Y.iloc[train_idx], Y.iloc[val_idx]

  # Buat dataset LightGBM
  dtrain = lgb.Dataset(X_tr, label=Y_tr, categorical_feature=cat_features)
  dval   = lgb.Dataset(X_val, label=Y_val, reference=dtrain)

  model = lgb.train(
      params,
      dtrain,
      num_boost_round=3000,          # Maksimal 3000 pohon
      valid_sets=[dval],
      callbacks=[
          lgb.early_stopping(150, verbose=False),  # Berhenti jika tidak ada perbaikan 150 iter
          lgb.log_evaluation(False)                 # Matikan log per iterasi
      ]
  )

  #Prediksi data validasi dan RMSE fold
  val_pred           = model.predict(X_val)
  oof_preds[val_idx] = val_pred
  rmse               = np.sqrt(mean_squared_error(Y_val, val_pred))
  scores.append(rmse)

  #prediksi test
  test_preds += model.predict(X_test) / n_splits

  print(f"  Fold {fold+1}: RMSE = {rmse:.4f}  |  Best iteration: {model.best_iteration}")

#Evaluasi Model
overall_rmse = np.sqrt(mean_squared_error(Y, oof_preds))
print(f"\n{'='*50}")
print(f"OOF RMSE    : {overall_rmse:.4f}")
print(f"Mean CV RMSE: {np.mean(scores):.4f} ± {np.std(scores):.4f}")
print(f"{'='*50}")

#15 fitur terpenting berdasarkan gain
imp = pd.DataFrame({
    'fitur'      : feature_cols,
    'importance' : model.feature_importance('gain')
})
imp = imp.sort_values('importance', ascending=False)
print("\n15 Fitur Terpenting:")
print(imp.head(15).to_string(index=False))


#Finalisasi
submission = pd.DataFrame({
    'sample_id'               : test['sample_id'],
    'property_organic_content': test_preds
})

submission.to_csv('submission.csv', index=False)

print(f"\nSubmission tersimpan! (submission.csv)")
print(f"Jumlah prediksi: {len(submission)}")
print("\nContoh hasil prediksi:")
print(submission.head(10))
print("\nStatistik prediksi:")
print(submission['property_organic_content'].describe().round(2))
```
